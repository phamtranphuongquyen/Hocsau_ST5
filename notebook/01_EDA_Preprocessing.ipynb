# Tiền xử lý dữ liệu & Feature Engineering

Các bước (đã sửa so với bản cũ):
1. Đọc lại dữ liệu gốc (độc lập với phần EDA phía trên).
2. Loại outlier `GrLivArea > 4000` & `SalePrice < 300000` (khuyến nghị của De Cock, 2011) **trước** khi điền missing.
3. Xóa 4 cột thiếu quá nhiều (`Alley`, `PoolQC`, `Fence`, `MiscFeature`).
4. Điền missing có chủ đích (NA = "không có tiện ích" -> `None`/0), phần còn lại dùng median/mode **tính trên tập train**.
5. Tạo đặc trưng mới, xử lý biến lệch (log1p), One-Hot Encoding.
6. Lưu dữ liệu sạch cho notebook 02 (sklearn) và 03 (PyTorch).

> Việc **đánh giá đặc trưng nào thực sự có ích** (ablation) nằm ở notebook `04_feature_experiments.ipynb`.

```python
# Đọc lại dữ liệu gốc để phần tiền xử lý không phụ thuộc vào các thay đổi ở phần EDA
train = pd.read_csv("../data/train.csv")
test = pd.read_csv("../data/test.csv")

print("Train:", train.shape, "| Test:", test.shape)
print("Missing ban đầu - Train:", train.isnull().sum().sum(), "| Test:", test.isnull().sum().sum())
```

## 1. Loại outlier

```python
# Nhà > 4000 sqft nhưng giá rất thấp (< $300,000) -> điểm bất thường (theo tài liệu De Cock)
outlier_mask = (train["GrLivArea"] > 4000) & (train["SalePrice"] < 300000)
print("Số outlier bị loại:", outlier_mask.sum(), "| Id:", train.loc[outlier_mask, "Id"].tolist())

train = train.loc[~outlier_mask].reset_index(drop=True)
print("Kích thước train sau khi loại outlier:", train.shape)
```

## 2. Xử lý Missing Values

```python
# 2.1 Xóa các cột thiếu > 70% dữ liệu
drop_cols = ["Alley", "PoolQC", "Fence", "MiscFeature"]
train = train.drop(columns=drop_cols)
test = test.drop(columns=drop_cols)

# 2.2 Cột categorical mà NA mang nghĩa "không có tiện ích" -> điền 'None'
none_cols = [
    "GarageType", "GarageFinish", "GarageQual", "GarageCond",
    "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2",
    "FireplaceQu", "MasVnrType",
]
# 2.3 Cột số liên quan Garage/Basement/MasVnr: NA = không có -> điền 0
zero_cols = [
    "GarageYrBlt", "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2", "BsmtUnfSF",
    "TotalBsmtSF", "BsmtFullBath", "BsmtHalfBath", "GarageCars", "GarageArea",
]
for df in (train, test):
    for col in none_cols:
        df[col] = df[col].fillna("None")
    for col in zero_cols:
        df[col] = df[col].fillna(0)

# 2.4 LotFrontage: điền median theo Neighborhood (tính trên train), phần còn lại điền median toàn train
lot_median = train.groupby("Neighborhood")["LotFrontage"].median()
for df in (train, test):
    df["LotFrontage"] = df["LotFrontage"].fillna(df["Neighborhood"].map(lot_median))
    df["LotFrontage"] = df["LotFrontage"].fillna(train["LotFrontage"].median())

# 2.5 Categorical còn thiếu (vd. Electrical ở train; MSZoning, KitchenQual... ở test) -> điền mode của train
cat_cols = [c for c in train.columns if not pd.api.types.is_numeric_dtype(train[c])]
for col in cat_cols:
    mode_value = train[col].mode()[0]
    for df in (train, test):
        df[col] = df[col].fillna(mode_value)

# 2.6 Numeric còn thiếu (nếu có) -> median của train
for col in test.select_dtypes(include=np.number).columns:
    med = train[col].median()
    for df in (train, test):
        df[col] = df[col].fillna(med)

print("Missing còn lại - Train:", train.isnull().sum().sum(), "| Test:", test.isnull().sum().sum())
assert train.isnull().sum().sum() == 0 and test.isnull().sum().sum() == 0
```

## 3. Tách nhãn & gộp train/test để xử lý đặc trưng đồng bộ

```python
# Nhãn: log(1 + SalePrice) -> phù hợp với metric RMSE trên log giá của cuộc thi
y_train = np.log1p(train["SalePrice"])
test_ids = test["Id"]

train_features = train.drop(columns=["Id", "SalePrice"])
test_features = test.drop(columns=["Id"])
n_train = len(train_features)

all_features = pd.concat([train_features, test_features], axis=0).reset_index(drop=True)
print("all_features:", all_features.shape, "| n_train =", n_train)
```

## 4. Feature Engineering

```python
# 4.1 Đặc trưng mới
all_features["TotalSF"] = all_features["TotalBsmtSF"] + all_features["1stFlrSF"] + all_features["2ndFlrSF"]
all_features["TotalBaths"] = (all_features["FullBath"] + 0.5 * all_features["HalfBath"]
                              + all_features["BsmtFullBath"] + 0.5 * all_features["BsmtHalfBath"])
# clip(lower=0): một số nhà được sửa/bán sau năm YrSold sẽ cho tuổi âm
all_features["HouseAge"] = (all_features["YrSold"] - all_features["YearBuilt"]).clip(lower=0)
all_features["RemodAge"] = (all_features["YrSold"] - all_features["YearRemodAdd"]).clip(lower=0)
all_features["HasGarage"] = (all_features["GarageArea"] > 0).astype(int)
all_features["HasBsmt"] = (all_features["TotalBsmtSF"] > 0).astype(int)
all_features["HasFireplace"] = (all_features["Fireplaces"] > 0).astype(int)

# 4.2 MSSubClass là MÃ loại nhà (20, 30, 60...) chứ không phải số đo -> coi là categorical
all_features["MSSubClass"] = all_features["MSSubClass"].astype(str)

print("Đã tạo 7 đặc trưng mới và chuyển MSSubClass sang categorical.")
```

```python
# 4.3 Giảm độ lệch (skewness) của các biến số liên tục bằng log1p
num_df = all_features.select_dtypes(include=np.number)
continuous_cols = [c for c in num_df.columns if num_df[c].nunique() > 20 and num_df[c].min() >= 0]

skewness = all_features[continuous_cols].skew().sort_values(ascending=False)
skewed_cols = skewness[skewness > 0.75].index.tolist()

print(f"{len(skewed_cols)} cột bị lệch phải (skew > 0.75) sẽ được log1p:")
print(skewness[skewed_cols].round(2).head(15))

all_features[skewed_cols] = np.log1p(all_features[skewed_cols])
print("\nSkew sau khi biến đổi (top 5):")
print(all_features[skewed_cols].skew().sort_values(ascending=False).round(2).head(5))
```

## 5. One-Hot Encoding & chia lại Train/Test

```python
# One-Hot trên toàn bộ (train + test) để hai tập có cùng bộ cột
cat_cols_all = [c for c in all_features.columns if not pd.api.types.is_numeric_dtype(all_features[c])]
all_features_encoded = pd.get_dummies(all_features, columns=cat_cols_all, dtype=int)

X_train = all_features_encoded.iloc[:n_train].copy().reset_index(drop=True)
X_test = all_features_encoded.iloc[n_train:].copy().reset_index(drop=True)
y_train = y_train.reset_index(drop=True)

assert X_train.shape[0] == len(y_train) and list(X_train.columns) == list(X_test.columns)
assert not X_train.isnull().any().any() and not X_test.isnull().any().any()
print("X_train:", X_train.shape, "| X_test:", X_test.shape)
```

```python
# Lưu dữ liệu sạch cho notebook 02 (sklearn) và 03 (PyTorch)
X_train.to_csv("../data/X_train_clean.csv", index=False)
X_test.to_csv("../data/X_test_clean.csv", index=False)
pd.DataFrame({"SalePrice": y_train}).to_csv("../data/y_train_clean.csv", index=False)
print("Đã lưu dữ liệu sạch vào thư mục ../data/")
```

- Loại 2 outlier theo khuyến nghị của tác giả bộ dữ liệu; nhãn dùng `log1p(SalePrice)`.
- Không còn giá trị thiếu; median/mode tính trên **train** nên không rò rỉ thông tin từ test.
- Thêm 7 đặc trưng mới (`TotalSF`, `TotalBaths`, `HouseAge`, `RemodAge`, `HasGarage`, `HasBsmt`, `HasFireplace`), coi `MSSubClass` là categorical, `log1p` các biến liên tục lệch phải.
